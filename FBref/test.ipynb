```python
def get_xg(soup):
    code = soup.find_all("div", class_="score_xg")
    try:
        xg_home = float(code[0].text)
        xg_away = float(code[1].text)
    except (IndexError, ValueError):
        xg_home, xg_away = "NaN", "NaN" 
    return pd.DataFrame([{"xg_home": xg_home, "xg_away": xg_away}])
```
